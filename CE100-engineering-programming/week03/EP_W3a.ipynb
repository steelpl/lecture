# 숫자·문자열·리스트

값의 자료형과 위치에 따라 연산 결과가 달라지는 이유 확인

## 숫자와 문자열

숫자 3+4와 문자열 "3"+"4"의 결과를 비교하시오.

```python
print(3 + 4)
print("3" + "4")
print(int("3") + int("4"))
```

## 문자열의 위치

s="Hello Python"에서 첫 글자와 마지막 글자, 앞의 다섯 글자를 각각 출력하시오.

```python
s = "Hello Python"
print(s[0])
print(s[-1])
print(s[:5])
```

## 리스트의 조회와 수정

a=[1,2,3]의 두 번째 값을 20으로 바꾸고 4를 추가하시오. 길이와 합계를 확인하시오.

```python
a = [1, 2, 3]
a[1] = 20
a.append(4)
print(a, len(a), sum(a))
```

같은 + 기호도 숫자에서는 덧셈, 문자열·리스트에서는 연결로 작동. AI에 코드를 요청할 때 입력이 어떤 자료형인지 함께 전달.
