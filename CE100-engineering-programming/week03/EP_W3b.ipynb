# Handling tuples in Python:

1. Accessing Tuple Elements

```python
# create a tuple
tup = (1, 2, 3, 4, 5)

# access the second element
print(tup[1])

# access the last element
print(tup[-1])
```

2. Tuple Unpacking: assigning each element of a tuple to a separate variable. This is called "tuple unpacking"

```python
# create a tuple
tup = (1, 2, 3)

# unpack the tuple
a, b, c = tup

# print the unpacked variables
print(a, b, c)
```

3. Converting Tuple to List

```python
# create a tuple
tup = (1, 2, 3)

# convert tuple to list
lst = list(tup)

# print the list
print(lst)
```

4. Concatenating Tuples

```python
# create two tuples
tup1 = (1, 2, 3)
tup2 = (4, 5, 6)

# concatenate the tuples
tup3 = tup1 + tup2

# print the concatenated tuple
print(tup3)
```

# Handling Dictionary in Python:

1. Creating a Dictionary

```python
# create a dictionary
my_dict = {"name": "John", "age": 30, "city": "New York"}

# print the dictionary
print(my_dict) 
```

2. Accessing Dictionary Values

```python
# create a dictionary
my_dict = {"name": "John", "age": 30, "city": "New York"}

# access the value of the "name" key
print(my_dict["name"])
```

3. Adding or Updating Dictionary Values

```python
# create a dictionary
my_dict = {"name": "John", "age": 30, "city": "New York"}

# add a new key-value pair
my_dict["email"] = "john@example.com"

# update the value of the "age" key
my_dict["age"] = 31

# print the updated dictionary
print(my_dict)
```

4. Removing Dictionary Elements

```python
# create a dictionary
my_dict = {"name": "John", "age": 30, "city": "New York"}

# remove the "age" key-value pair
del my_dict["age"]

# print the updated dictionary
print(my_dict)
```

# Handling Set in Python:




1. Creating a Set

```python
# create a set using the set() function
my_set = set([1, 2, 3, 4, 5])

# print the set
print(my_set)

# create a set using curly braces
my_set = {1, 2, 3, 4, 5}

# print the set
print(my_set)
```

2. Adding Elements to a Set

```python
# create a set
my_set = {1, 2, 3}

# add an element to the set
my_set.add(4)

# print the updated set
print(my_set)
```

3. Removing Elements from a Set

```python
# create a set
my_set = {1, 2, 3, 4}

# remove an element from the set
my_set.remove(3)

# print the updated set
print(my_set)
```

4. Set Operations

```python
# create two sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}

# union of two sets
union_set = set1 | set2

# intersection of two sets
intersection_set = set1 & set2

# difference of two sets
difference_set = set1 - set2

# print the results
print(union_set)         
print(intersection_set)  
print(difference_set)    
```

# Handling Bool in Python:

1. Creating a Boolean Variable

```python
# create a Boolean variable
my_bool = True

# print the Boolean variable
print(my_bool)
```

2. Boolean Operators

```python
# Boolean AND
print(True and True)
print(True and False)
print(False and False)

# Boolean OR
print(True or True)
print(True or False)
print(False or False)

# Boolean NOT
print(not True)
print(not False)
```

3. Boolean Comparisons

```python
# greater than
print(5 > 3)

# less than
print(3 < 5)

# equal to
print(3 == 3)

# not equal to
print(3 != 5)
```

# Handling variables in Python:

1. Assigning Values to Variables

```python
# assign a value to a variable
x = 5

# print the value of the variable
print(x)
```

2. Changing the Value of a Variable

```python
# assign a value to a variable
x = 5

# change the value of the variable
x = 10

# print the value of the variable
print(x)
```

3. Multiple Assignment

```python
# assign values to multiple variables
x, y, z = 1, 2, 3

# print the values of the variables
print(x)  # Output: 1
print(y)  # Output: 2
print(z)  # Output: 3
```
