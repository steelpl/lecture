2. Tuple Unpacking: assigning each element of a tuple to a separate variable. This is called "tuple unpacking"


```python
# create a tuple
tup = (1, 2, 3)

# unpack the tuple
a, b, c = tup

# print the unpacked variables
print(a, b, c)
```

3. Converting Tuple to List


```python
# create a tuple
tup = (1, 2, 3)

# convert tuple to list
lst = list(tup)

# print the list
print(lst)
```

4. Concatenating Tuples


```python
# create two tuples
tup1 = (1, 2, 3)
tup2 = (4, 5, 6)

# concatenate the tuples
tup3 = tup1 + tup2

# print the concatenated tuple
print(tup3)
```

```python
assert tup3 == (1, 2, 3, 4, 5, 6)
assert isinstance(tup, tuple)
print("원본 예제 확인 완료")
```

# Handling Set in Python:


1. Creating a Set


```python
# create a set using the set() function
my_set = set([1, 2, 3, 4, 5])

# print the set
print(my_set)

# create a set using curly braces
my_set = {1, 2, 3, 4, 5}

# print the set
print(my_set)
```

2. Adding Elements to a Set


```python
# create a set
my_set = {1, 2, 3}

# add an element to the set
my_set.add(4)

# print the updated set
print(my_set)
```

3. Removing Elements from a Set


```python
# create a set
my_set = {1, 2, 3, 4}

# remove an element from the set
my_set.remove(3)

# print the updated set
print(my_set)
```

4. Set Operations


```python
# create two sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}

# union of two sets
union_set = set1 | set2

# intersection of two sets
intersection_set = set1 & set2

# difference of two sets
difference_set = set1 - set2

# print the results
print(union_set)         
print(intersection_set)  
print(difference_set)    
```

```python
assert intersection_set == {3}
assert difference_set == {1, 2}
print("원본 예제 확인 완료")
```
