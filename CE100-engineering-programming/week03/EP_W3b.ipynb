## 1. 먼저 원리를 읽기

집합에는 리스트처럼 위치 번호가 없습니다. `True`/`False`는 조건의 결과이며, 변수는 값을 참조하는 이름입니다.

`stations`의 길이와 `unique_stations`의 길이는 각각 얼마일까요? 중복을 없앤 결과에서 첫 번째 관측 지점을 복원할 수 있을까요?

- 그렇게 생각한 이유:
- 실행 뒤 발견한 차이:


````python
stations = ["A", "B", "A", "C"]
unique_stations = set(stations)
location = (37.5, 127.0)  # 위도, 경도라는 순서를 약속한 튜플
rain_mm = {"A": 0.0, "B": 12.0, "C": 3.0}
print("전체 관측 수:", len(stations))
print("서로 다른 지점 수:", len(unique_stations))
print("B 지점 강수량:", rain_mm["B"], "mm")
assert len(stations) == 4 and len(unique_stations) == 3
assert rain_mm["B"] >= 10
````

### 관찰을 원리와 연결하기

중복 제거와 순서 보존은 서로 다른 요구입니다. `set`으로 바꾸면 관측 횟수 정보도 사라집니다. 지점 `A`의 0 mm는 관측값이고, 키가 없는 `D`와는 다릅니다.

````python
print("A:", rain_mm.get("A"), "D:", rain_mm.get("D"))
first_seen = list(dict.fromkeys(stations))
print("처음 관측된 순서:", first_seen)
assert first_seen == ["A", "B", "C"]
# 예측한 뒤 stations의 순서를 바꾸고 다시 실행해 보세요.
print("좌표:", location, "자료형:", type(location).__name__)
````
