## 1. 먼저 원리를 읽기

예외는 계약을 수행할 수 없다는 신호입니다. `try`에는 실패할 수 있는 최소 작업을 두고, 처리할 수 있는 예외만 잡습니다.

## 2. 실행 전 예측하기

`float(" 2.5 ")`, `float("rain")`, `float("nan")` 중 어떤 것은 예외 없이 숫자로 변환되지만 관측값으로 사용할 수 없을까요?


아래 기록을 먼저 채운 뒤 다음 코드 셀을 실행하세요. 정답과 다르면 예측에 사용한 가정을 고칩니다.

- 내 예상 결과:
- 그렇게 생각한 이유:
- 실행 뒤 발견한 차이:


````python
import math
for text in [" 2.5 ", "rain", "nan"]:
    try:
        value = float(text)
        print(repr(text), "변환:", value, "유한한가:", math.isfinite(value))
    except ValueError as error:
        print(repr(text), "변환 실패:", type(error).__name__)
assert not math.isfinite(float("nan"))
````

### 관찰을 원리와 연결하기

변환 성공은 데이터 유효성을 보장하지 않습니다. `NaN`과 무한대는 유한하지 않습니다. 구체적인 실패 이유가 있어야 사용자가 입력을 고치거나 데이터를 제외할 수 있습니다.


다음 셀의 입력이나 조건을 하나 바꾸세요. 바꾼 값, 예상한 변화, 실제 결과를 적습니다. 단순 재실행만으로 이 활동을 마치지 않습니다.


````python
def divide(a, b):
    if b == 0:
        raise ValueError("분모는 0일 수 없습니다")
    return a / b

assert divide(6, 2) == 3
try:
    divide(6, 0)
except ValueError as error:
    print("예상한 오류를 확인했습니다:", error)
````

## 4. AI 제안을 검토하고 수정 지시하기

````python
def ai_suggestion(text):
    try:
        return float(text)
    except ValueError:
        return 0.0

print("무강수:", ai_suggestion("0"), "입력 오류:", ai_suggestion("rain"))
assert ai_suggestion("rain") == ai_suggestion("0")
````
