# 조건과 반복

조건으로 대상을 고르고 반복문으로 계산을 누적

## 조건문

점수 75가 60 이상이면 통과를 출력하시오. 점수를 55로 바꾸어 결과를 비교하시오.

```python
score = 75
if score >= 60:
    print("통과")
else:
    print("재확인")
```

## for 반복문

1부터 10까지의 합을 구하시오. 각 반복의 total을 출력하여 값이 쌓이는 과정을 확인하시오.

```python
total = 0
for value in range(1, 11):
    total += value
    print(value, total)
```

## while 반복문

반복을 끝내는 조건을 찾아보시오. 조건을 value<5로 바꾸어 for 예제의 일부 결과와 비교하시오.

```python
value = 1
while value <= 5:
    print(value)
    value += 1
```

## 조건과 반복 함께 사용

values에서 양수만 더하시오. >를 >=로 바꾸었을 때 합계가 달라지는지 설명하시오.

```python
values = [-2, 0, 3, 5]
total = 0
for value in values:
    if value > 0:
        total += value
print(total)
```
