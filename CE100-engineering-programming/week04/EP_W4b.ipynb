## 2. 실행 전 예측하기

`show_total([2, 4])`를 변수에 대입하면 그 변수에는 6과 `None` 중 무엇이 들어갈까요?


아래 기록을 먼저 채운 뒤 다음 코드 셀을 실행하세요. 정답과 다르면 예측에 사용한 가정을 고칩니다.

- 내 예상 결과:
- 그렇게 생각한 이유:
- 실행 뒤 발견한 차이:


````python
def show_total(values):
    print(sum(values))

def get_total(values):
    return sum(values)

shown = show_total([2, 4])
returned = get_total([2, 4])
print("shown:", shown, "returned:", returned)
assert shown is None and returned == 6
````

### 관찰을 원리와 연결하기

함수 안에서 값을 계산해도 `return`이 없으면 반환값은 `None`입니다. 다른 계산에 사용할 값인지, 화면에 보여 줄 문장인지 먼저 구분하세요.


다음 셀의 입력이나 조건을 하나 바꾸세요. 바꾼 값, 예상한 변화, 실제 결과를 적습니다. 단순 재실행만으로 이 활동을 마치지 않습니다.


````python
def count_wet_days(values, threshold=1.0):
    return sum(value >= threshold for value in values)

print(count_wet_days([0, 1, 2]))
print(count_wet_days([0, 1, 2], threshold=2))
assert count_wet_days([0, 1, 2]) == 2
# threshold를 0으로 바꾸면 어떤 날이 포함되는지 먼저 예측하세요.
````

````python
def check_answer(candidate):
    import math
    for values, expected in [([0, 2, 4], 2), ([3], 3), ([0, 0], 0)]:
        before = values.copy()
        assert math.isclose(candidate(values), expected), (values, expected)
        assert values == before
    for values in [[], [2, -1]]:
        try:
            candidate(values)
        except ValueError:
            pass
        else:
            raise AssertionError(f"{values}는 ValueError여야 합니다")
    print("공개 점검 통과: 반환값·평균·빈 입력·음수")

try:
    check_answer(mean_rainfall)
except NotImplementedError:
    print("미완료: 위 TODO를 구현한 뒤 이 셀을 다시 실행하세요. 아직 점검을 통과하지 않았습니다.")
````
