## 1. 먼저 원리를 읽기

배열의 숫자만으로는 행·열의 의미를 알 수 없습니다. 여기서는 **행=지점, 열=날짜**, 단위는 mm입니다.
`axis=0`은 행 방향으로 모아 날짜별 값을, `axis=1`은 열 방향으로 모아 지점별 값을 구합니다. `*`는 원소별 곱이고 `@`는 행렬곱입니다. reshape는 값의 의미까지 자동으로 바꾸지 않습니다.


## 2. 실행 전 예측하기

2개 지점×3일 강수량에서 지점별 합계의 shape는 무엇일까요? 일별 합계와 값도 먼저 계산하세요.


아래 기록을 먼저 채운 뒤 다음 코드 셀을 실행하세요. 정답과 다르면 예측에 사용한 가정을 고칩니다.

- 내 예상 결과:
- 그렇게 생각한 이유:
- 실행 뒤 발견한 차이:


````python
import numpy as np

rain = np.array([[0.0, 2.0, 4.0], [1.0, 3.0, 5.0]])
print("shape:", rain.shape)
print("지점별 합계:", rain.sum(axis=1))
print("날짜별 합계:", rain.sum(axis=0))
np.testing.assert_allclose(rain.sum(axis=1), [6, 9])
np.testing.assert_allclose(rain.sum(axis=0), [1, 5, 9])
````

### 관찰을 원리와 연결하기

정방행렬에서는 잘못된 축으로 집계해도 출력 길이가 같아 오류를 놓치기 쉽습니다. 행·열 개수가 다른 예제를 먼저 시험하세요.


다음 셀의 입력이나 조건을 하나 바꾸세요. 바꾼 값, 예상한 변화, 실제 결과를 적습니다. 단순 재실행만으로 이 활동을 마치지 않습니다.


````python
converted = rain / 1000  # mm -> m; 원본은 바꾸지 않음
assert rain[0, 1] == 2 and converted[0, 1] == 0.002
mask = rain >= 3
print("3 mm 이상:", rain[mask])
# 배열 슬라이스는 원본과 메모리를 공유할 수 있습니다.
independent = rain[:, :2].copy()
independent[0, 0] = 99
assert rain[0, 0] == 0
print("원소별 제곱:", rain * rain)
````
