# 배열 생성


## 리스트에서 행렬/배열 생성


```python
import numpy as np      # NumPy 라이브러리를 np라는 이름으로 불러옴

a = [[1,2,3], [4,5,6]]  # 2차원 배열을 나타내는 파이썬 리스트 생성
b = np.array(a)         # 파이썬 리스트 a를 NumPy 배열로 변환하여 b에 저장
print(b)                # NumPy 배열 b를 출력
```

```python
print(b.ndim)  # NumPy 배열 b의 차원수
```

```python
print(b.shape)  # 배열의 각 차원의 크기 (튜플 형태) 출력
```

```python
print(b[0, 0])  # 배열의 첫 번째 행, 첫 번째 열의 원소 (인덱스 [0, 0]) 출력
```

# Numpy 연산
* 배열간 연산
* `+,-,*,/` 등의 연산자 사용


```python
import numpy as np  # NumPy 라이브러리 불러오기

a = np.array([1, 2, 3])  # 배열 a 생성
b = np.array([4, 5, 6])  # 배열 b 생성

# 각 요소 더하기
c= np.add(a, b)
print(c)
```

```python
# 각 요소 곱하기
c = np.multiply(a, b)
print(c)
```

```python
# 각 요소 나누기
c = np.divide(a, b)
print(c)
```

```python
arr1 = [[1, 2], [3, 4]]   # 2차원 배열(리스트) 생성
arr2 = [[5, 6], [7, 8]]   # 2차원 배열(리스트) 생성
a = np.array(arr1)        # NumPy 배열로 변환
b = np.array(arr2)        # NumPy 배열로 변환

# 두 배열의 행렬 곱셈
c = np.dot(a, b)
print(c)
```

```python
a = np.array([[-1, 2, 3], [3, 4, 8]])   # 배열 a 생성
s = np.sum(a)                           # 배열 a의 모든 요소 합
print('sum=', a.sum())

print('sum by column=', a.sum(axis=0))  # 행 축을 줄여 각 열의 합
print('sum by row=', a.sum(axis=1))     # 열 축을 줄여 각 행의 합
print('mean=', a.mean())                # 평균
print('sd=', a.std())                   # 표준편차
print('product=', a.prod())             # 모든 요소의 곱 출력
```

```python
np.testing.assert_array_equal(a.sum(axis=0), [2, 6, 11])
np.testing.assert_array_equal(a.sum(axis=1), [4, 15])
print("예제 계산 확인")
```

## 선택 확장 · 수학 함수·선형대수

각 함수가 반환하는 자료형·shape와 필요한 입력 조건 먼저 확인.

# 기본 수학함수


| 함수 | 설명 |
|---|---|
| `abs`, `fabs` | 정수, 부동 소수점 또는 복소수 값에 대한 요소별 절대값 |
| `sqrt` | 각 요소의 제곱근 ( `arr ** 0.5` 와 동일) |
| `square` | 각 요소의 제곱 ( `arr ** 2` 와 동일) |
| `exp` | 각 요소의 지수 (e<sup>x</sup>) |
| `log`, `log10`, `log2`, `log1p` | 각각 자연 로그 (밑 e), 밑 10 로그, 밑 2 로그 및 log(1 + x) |
| `sign` | 각 요소의 부호: 1 (양수), 0 (0), 또는 -1 (음수) |
| `ceil` | 각 요소의 올림 (즉, 주어진 수보다 크거나 같은 최소 정수) |
| `floor` | 각 요소의 내림 (즉, 주어진 수보다 작거나 같은 최대 정수) |
| `rint` | 반올림 |
| `isnan` | NaN (숫자가 아님) 확인, `[True|False]` |
| `isfinite`, `isinf` | 유한/무한 확인 |
| `cos`, `cosh`, `sin`, `sinh`, `tan`, `tanh` | 삼각 함수 |
| `arccos`, `arccosh`, `arcsin`, `arcsinh`, `arctan`, `arctanh` | 역삼각 함수 |


```python
import numpy as np

# 절대값
arr = np.array([-1, 2, -3])
abs_arr = np.abs(arr)  # 또는 np.fabs(arr)
print("abs:", abs_arr)
```

```python
# 제곱근
arr = np.array([4, 9, 16])
sqrt_arr = np.sqrt(arr)
print("sqrt:", sqrt_arr)
```

```python
# 제곱
arr = np.array([2, 3, 4])
square_arr = np.square(arr)
print("square:", square_arr)
```

```python
# 지수
arr = np.array([1, 2, 3])
exp_arr = np.exp(arr)
print("exp:", exp_arr)
```

```python
# 로그
arr = np.array([1, 10, 100])
log_arr = np.log(arr)       # 자연 로그
log10_arr = np.log10(arr)   # 밑 10 로그
log2_arr = np.log2(arr)     # 밑 2 로그
log1p_arr = np.log1p(arr)   # log(1 + x)
print("log:", log_arr)
print("log10:", log10_arr)
print("log2:", log2_arr)
print("log1p:", log1p_arr)
```

```python
# 부호
arr = np.array([-2, 0, 3])
sign_arr = np.sign(arr)
print("sign:", sign_arr)
```

```python
# 올림
arr = np.array([1.2, 2.7, 3.5])
ceil_arr = np.ceil(arr)
print("ceil:", ceil_arr)
```

```python
# 내림
arr = np.array([1.2, 2.7, 3.5])
floor_arr = np.floor(arr)
print("floor:", floor_arr)
```

```python
# 반올림
arr = np.array([1.2, 2.7, 3.5])
rint_arr = np.rint(arr)
print("rint:", rint_arr)
```

```python
# NaN 확인
arr = np.array([1, np.nan, 3])
isnan_arr = np.isnan(arr)
print("isnan:", isnan_arr)
```

```python
# 유한/무한 확인
arr = np.array([1, np.inf, 3])
isfinite_arr = np.isfinite(arr)
isinf_arr = np.isinf(arr)
print("isfinite:", isfinite_arr)
print("isinf:", isinf_arr)
```

```python
# 삼각 함수 (예: sin)
arr = np.array([0, np.pi/2, np.pi])
sin_arr = np.sin(arr)
print("sin:", sin_arr)
```

```python
# 역삼각 함수 (예: arcsin)
arr = np.array([0, 1, -1])
arcsin_arr = np.arcsin(arr)
print("arcsin:", arcsin_arr)
```

# 선형대수(벡터/행렬연산) 함수


| 함수 | 설명 |
|---|---|
| `diag` | 정방 행렬의 대각선 원소를 1차원 배열로 반환하거나, 1차원 배열을 대각선 이외의 요소가 0인 정방 행렬로 변환 |
| `dot` | 행렬 곱셈 |
| `trace` | 대각선 원소의 합 |
| `det` | 행렬식 |
| `eig` | 정방 행렬의 고유값과 고유 벡터 |
| `inv` | 정방 행렬의 역행렬 |
| `pinv` | 행렬의 Moore-Penrose 유사 역행렬 |
| `qr` | QR 분해 |
| `svd` | 특이값 분해 (SVD) |
| `solve` | Ax = b에서 x 해를 구함. 여기서 A는 행렬 |
| `lstsq` | Ax = b에 대한 최소 제곱 해 |


```python
import numpy as np

# diag: 대각선 원소 추출 또는 대각 행렬 생성
a = np.array([[1, 2], [3, 4]])
d = np.diag(a)  # 대각선 원소 추출
print("diag(a):", d)
d = np.diag(d)  # 대각 행렬 생성
print("diag(d):", d)
```

```python
# dot: 행렬 곱셈
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
c = np.dot(a, b)  # 행렬 곱셈
print("dot(a, b):", c)
```

```python
# trace: 대각선 원소의 합
a = np.array([[1, 2], [3, 4]])
t = np.trace(a)  # 대각선 원소의 합
print("trace(a):", t)
```

```python
# det: 행렬식
a = np.array([[1, 2], [3, 4]])
d = np.linalg.det(a)  # 행렬식
print("det(a):", d)
```

```python
# eig: 고유값과 고유벡터
a = np.array([[1, 2], [3, 4]])
w, v = np.linalg.eig(a)  # 고유값(w)과 고유벡터(v)
print("eig(a):", w, v)
```

```python
# inv: 역행렬
a = np.array([[1, 2], [3, 4]])
inv_a = np.linalg.inv(a)  # 역행렬
print("inv(a):", inv_a)
```

```python
# pinv: Moore-Penrose 유사 역행렬
a = np.array([[1, 2], [3, 4]])
pinv_a = np.linalg.pinv(a)  # 유사 역행렬
print("pinv(a):", pinv_a)
```

```python
# qr: QR 분해
a = np.array([[1, 2], [3, 4]])
q, r = np.linalg.qr(a)  # QR 분해
print("qr(a):", q, r)
```

```python
# svd: 특이값 분해 (SVD)
a = np.array([[1, 2], [3, 4]])
u, s, vh = np.linalg.svd(a)  # 특이값 분해
print("svd(a):", u, s, vh)
```

```python
# solve: Ax = b에서 x 해 구하기
a = np.array([[1, 2], [3, 4]])
b = np.array([5, 6])
x = np.linalg.solve(a, b)  # x 해
print("solve(a, b):", x)
```

```python
# lstsq: 최소 제곱 해
a = np.array([[1, 2], [3, 4]])
b = np.array([5, 6])
x, residuals, rank, s = np.linalg.lstsq(a, b, rcond=None)  # 최소 제곱 해
print("lstsq(a, b):", x, residuals, rank, s)
```

```python
np.testing.assert_allclose(a @ x, b)
print("예제 계산 확인")
```

### 이해 확인·변형

## 선택 확장 · 난수 분포

각 분포의 모수와 출력 범위 기록. 매번 표본 평균이 정확히 같아야 하는가?

# 난수 생성


| 함수 | 분포 |
|---|---|
| `beta(a, b[, size])` | 베타 분포 |
| `binomial(n, p[, size])` | 이항 분포 |
| `chisquare(df[, size])` | 카이 제곱 분포 |
| `exponential([scale, size])` | 지수 분포 |
| `f(dfnum, dfden[, size])` | F 분포 |
| `gamma(shape[, scale, size])` | 감마 분포 |
| `geometric(p[, size])` | 기하 분포 |
| `hypergeometric(ngood, nbad, nsample[, size])` | 초기하 분포 |
| `multinomial(n, pvals[, size])` | 다항 분포 |
| `negative_binomial(n, p[, size])` | 음이항 분포 |
| `normal([loc, scale, size])` | 정규 (가우시안) 분포 |
| `poisson([lam, size])` | 포아송 분포 |


```python
import numpy as np
import matplotlib.pyplot as plt

# 베타 분포
s = np.random.beta(a=5, b=3, size=1000)
print("beta: mean={:.5f}, std={:.5f}".format(np.mean(s), np.std(s)))
plt.hist(s, bins=50)
plt.title("Beta Distribution")
plt.show()
```

```python
# 이항 분포
s = np.random.binomial(n=10, p=0.5, size=1000)
print("binomial: mean={:.5f}, std={:.5f}".format(np.mean(s), np.std(s)))
plt.hist(s, bins=10)  # bins는 발생 가능한 값의 개수에 맞게 조정
plt.title("Binomial Distribution")
plt.show()
```

```python
# 카이 제곱 분포
s = np.random.chisquare(df=2, size=1000)
print("chisquare: mean={:.5f}, std={:.5f}".format(np.mean(s), np.std(s)))
plt.hist(s, bins=50)
plt.title("Chi-Square Distribution")
plt.show()
```

```python
# 지수 분포
s = np.random.exponential(scale=1.0, size=1000)
print("exponential: mean={:.5f}, std={:.5f}".format(np.mean(s), np.std(s)))
plt.hist(s, bins=50)
plt.title("Exponential Distribution")
plt.show()
```

```python
# F 분포
s = np.random.f(dfnum=3, dfden=5, size=1000)
print("f: mean={:.5f}, std={:.5f}".format(np.mean(s), np.std(s)))
plt.hist(s, bins=50)
plt.title("F Distribution")
plt.show()
```

```python
# 감마 분포
s = np.random.gamma(shape=2.0, scale=1.0, size=1000)
print("gamma: mean={:.5f}, std={:.5f}".format(np.mean(s), np.std(s)))
plt.hist(s, bins=50)
plt.title("Gamma Distribution")
plt.show()
```

```python
# 기하 분포
s = np.random.geometric(p=0.5, size=1000)
print("geometric: mean={:.5f}, std={:.5f}".format(np.mean(s), np.std(s)))
plt.hist(s, bins=10)  # bins는 발생 가능한 값의 개수에 맞게 조정
plt.title("Geometric Distribution")
plt.show()
```

```python
# 초기하 분포
s = np.random.hypergeometric(ngood=10, nbad=15, nsample=5, size=1000)
print("hypergeometric: mean={:.5f}, std={:.5f}".format(np.mean(s), np.std(s)))
plt.hist(s, bins=10)  # bins는 발생 가능한 값의 개수에 맞게 조정
plt.title("Hypergeometric Distribution")
plt.show()
```

```python
# 다항 분포
s = np.random.multinomial(n=10, pvals=[0.2, 0.3, 0.5], size=1000)
print("multinomial:", s)
```

```python
# 음이항 분포
s = np.random.negative_binomial(n=5, p=0.5, size=1000)
print("negative_binomial: mean={:.5f}, std={:.5f}".format(np.mean(s), np.std(s)))
plt.hist(s, bins=50)
plt.title("Negative Binomial Distribution")
plt.show()
```

```python
# 정규 (가우시안) 분포
s = np.random.normal(loc=0.0, scale=1.0, size=1000)
print("normal: mean={:.5f}, std={:.5f}".format(np.mean(s), np.std(s)))
plt.hist(s, bins=50)
plt.title("Normal Distribution")
plt.show()
```

```python
# 포아송 분포
s = np.random.poisson(lam=3.0, size=1000)
print("poisson: mean={:.5f}, std={:.5f}".format(np.mean(s), np.std(s)))
plt.hist(s, bins=10)  # bins는 발생 가능한 값의 개수에 맞게 조정
plt.title("Poisson Distribution")
plt.show()
```

seed를 고정하고 표본 수를 바꿔 분포의 모양·표본 통계의 변화 확인
