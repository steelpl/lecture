## 2. 실행 전 예측하기

A에만 2 mm를 추가한 뒤 B의 합계는 얼마일까요? 클래스 정의 안에 `readings = []`를 두면 어떤 문제가 생길까요?


아래 기록을 먼저 채운 뒤 다음 코드 셀을 실행하세요. 정답과 다르면 예측에 사용한 가정을 고칩니다.

- 내 예상 결과:
- 그렇게 생각한 이유:
- 실행 뒤 발견한 차이:


````python
class RainGauge:
    unit = "mm"

    def __init__(self, name):
        self.name = name
        self.readings = []  # 객체마다 새로운 리스트

    def add(self, value):
        self.readings.append(value)

    def total(self):
        return sum(self.readings)

gauge_a = RainGauge("A")
gauge_b = RainGauge("B")
gauge_a.add(2)
print(gauge_a.name, gauge_a.total(), gauge_b.name, gauge_b.total())
assert gauge_a.total() == 2 and gauge_b.total() == 0
````

### 관찰을 원리와 연결하기

인스턴스 속성은 객체마다, 클래스 속성은 클래스에 속합니다. `unit`은 공유해도 되지만 수정 가능한 관측 리스트까지 공유하면 지점별 결과가 섞입니다.


다음 셀의 입력이나 조건을 하나 바꾸세요. 바꾼 값, 예상한 변화, 실제 결과를 적습니다. 단순 재실행만으로 이 활동을 마치지 않습니다.


````python
gauge_b.add(7)
gauge_a.add(3)
assert gauge_a.readings == [2, 3]
assert gauge_b.readings == [7]
assert gauge_a.readings is not gauge_b.readings
print("A의 총강수량:", gauge_a.total(), gauge_a.unit)
````
