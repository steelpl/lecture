# Definition of Class and Object

>
* A **class** is a blueprint or a template for creating objects. It defines a set of attributes and methods that an object of that class will have.
* An **object** is an instance of a class. It is created based on the blueprint defined by the class.


# An intuitive example

```python
# we define a class Person that has two attributes name and age
class Person:
    # initializing the attributes of an object when it is created.
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # a method say_hello that prints a greeting message using the name attribute.
    def say_hello(self):
        print("Hello, my name is", self.name, "and I am", self.age, "years old")
```

```python
# Creating objects of this class
p1 = Person("Alice", 25)
p2 = Person("Tom", 21)
```

```python
# Calling the say_hello method on the object:
p1.say_hello()
p2.say_hello()
```

```python
class FourCal:
  def __init__(self, first, second):
    self.first = first
    self.second = second
  def add(self):
    result = self.first + self.second
    return result

  def mul(self):
    result = self.first * self.second
    return result

  def sub(self):
    result = self.first - self.second
    return result

  def div(self):
    result = self.first / self.second
    return result
```

# Class for arithmatic operations

```python
class FourCal:
  def __init__(self, first, second):
    self.first = first
    self.second = second

  def add(self):
    result = self.first + self.second
    return result

  def mul(self):
    result = self.first * self.second
    return result

  def sub(self):
    result = self.first - self.second
    return result

  def div(self):
    result = self.first / self.second
    return result
```

```python
a = FourCal(4,2)

print(a.add())
print(a.mul())
print(a.sub())
print(a.div())
```

```python
# inheritance
class MoreFourCal(FourCal):
  pass
```

```python
a = MoreFourCal(4,2)
a.mul()
```

# Examples of Inheritances

```python
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species

    def eat(self):
        print(f"{self.name} is eating")

    def sleep(self):
        print(f"{self.name} is sleeping")

class Dog(Animal):
    def __init__(self, name, species, breed):
        super().__init__(name, species)
        self.breed = breed

    def bark(self):
        print("Woof!")

my_dog = Dog("Fido", "Canis lupus familiaris", "Golden Retriever")
my_dog.eat()
my_dog.bark()
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name} and I am {self.age} years old")

class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id

    def study(self):
        print(f"{self.name} is studying")

my_student = Student("John", 20, "1234")
my_student.introduce()
my_student.study()
```

```python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model

    def start(self):
        print("Starting the vehicle")

class Car(Vehicle):
    def __init__(self, make, model, year):
        super().__init__(make, model)
        self.year = year

    def start(self):
        print("Starting the car")
        print(f"my car is made in {self.year}")

my_car = Car("Honda", "Civic", 2022)
my_car.start()
```

# Class variable

```python
class Car:
    num_wheels = 4

    def __init__(self, make, model):
        self.make = make
        self.model = model

my_car = Car("Honda", "Civic")
print(my_car.num_wheels)
print(Car.num_wheels)
```

```python
Car.num_wheels = 10
my_car = Car("Honda", "Civic")
print(my_car.num_wheels)
print(Car.num_wheels)
```

```python
print(id(my_car.num_wheels))
print(id(Car.num_wheels))
```
