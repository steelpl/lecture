## 1. 먼저 원리를 읽기

파일 작업은 **어디에(경로), 어떤 문자 규칙으로(인코딩), 어떤 방식으로(모드)** 읽고 쓰는지 정해야 합니다.

## 2. 실행 전 예측하기

`w`로 "가"를 쓴 뒤 `a`로 "나"를 추가하면 어떤 문자열이 남을까요? 두 번째도 `w`라면요?


아래 기록을 먼저 채운 뒤 다음 코드 셀을 실행하세요. 정답과 다르면 예측에 사용한 가정을 고칩니다.

- 내 예상 결과:
- 그렇게 생각한 이유:
- 실행 뒤 발견한 차이:


````python
from pathlib import Path
import tempfile

workspace = Path(tempfile.mkdtemp(prefix="ce100-io-"))
source = workspace / "newfile.txt"
source.write_text("가\n", encoding="utf-8")
with source.open("a", encoding="utf-8") as file:
    file.write("나\n")
actual = source.read_text(encoding="utf-8")
print("연습 폴더:", workspace)
print(repr(actual))
assert actual == "가\n나\n"
````

````python
lines = source.read_text(encoding="utf-8").splitlines()
print("줄 목록:", lines, "줄 수:", len(lines))
replacement = workspace / "replace-demo.txt"
replacement.write_text("first\n", encoding="utf-8")
replacement.write_text("second\n", encoding="utf-8")
assert replacement.read_text(encoding="utf-8") == "second\n"
assert source.read_text(encoding="utf-8") == "가\n나\n"
````

## 4. AI 제안을 검토하고 수정 지시하기

다음은 오류를 검토하기 위해 만든 **가상 AI 답변**입니다. 실제 AI 응답이나 정답이 아닙니다.
실행 전에 어떤 요구사항을 놓쳤는지 찾고, 반례가 그 문제를 어떻게 드러내는지 설명하세요.


````python
def ai_suggestion(path, values):
    with path.open("a", encoding="utf-8") as file:
        for value in values:
            file.write(f"{value}\n")
    return len(values)

bad_path = workspace / "append-demo.txt"
bad_path.write_text("", encoding="utf-8")
ai_suggestion(bad_path, [1, 2])
ai_suggestion(bad_path, [1, 2])
print(repr(bad_path.read_text(encoding="utf-8")))
assert len(bad_path.read_text(encoding="utf-8").splitlines()) == 4
````
