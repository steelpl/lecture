## 2. 실행 전 예측하기

시간순으로 정렬되지 않은 관측을 그대로 선으로 이으면 무엇이 달라질까요? 날짜와 값의 연결을 따로 정렬해도 될까요?

- 내 예상 결과:
- 그렇게 생각한 이유:
- 실행 뒤 발견한 차이:


````python
import numpy as np
import matplotlib.pyplot as plt

days = np.array([3, 1, 2])
rain_mm = np.array([8., 0., 4.])
order = np.argsort(days)
fig, ax = plt.subplots(figsize=(6, 3))
line, = ax.plot(days[order], rain_mm[order], marker="o")
ax.set(xlabel="Day", ylabel="Rainfall (mm)", title="Synthetic daily rainfall")
ax.set_ylim(bottom=0)
plt.show()
np.testing.assert_array_equal(line.get_ydata(), [0, 4, 8])
assert ax.get_ylabel() == "Rainfall (mm)"
````

### 관찰을 원리와 연결하기

````python
fig2, axes = plt.subplots(1, 2, figsize=(8, 3))
for axis in axes:
    axis.bar(["A", "B"], [98, 100])
    axis.set_ylabel("Synthetic level (m)")
axes[0].set_ylim(0, 110)
axes[1].set_ylim(97, 101)
axes[0].set_title("Baseline 0")
axes[1].set_title("Zoomed axis")
fig2.tight_layout()
plt.show()
assert axes[0].patches[0].get_height() == axes[1].patches[0].get_height() == 98
````

### 공개 점검

아래 점검 셀을 실행합니다. **“미완료”는 통과가 아닙니다.** 위 함수를 구현한 뒤 다시 실행하세요.
