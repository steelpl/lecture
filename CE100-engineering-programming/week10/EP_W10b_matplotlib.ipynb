# sine 그래프 그리기


```python
# 필요한 라이브러리 임포트
import matplotlib.pyplot as plt
import numpy as np

# 그래프 스타일 설정
plt.rcParams["font.size"] = 12          # 폰트 크기 설정
plt.rcParams["figure.figsize"] = (8,6)  # 그래프 크기 설정

# x 데이터 생성
# -2π부터 2π까지 100개의 점을 생성
x = np.linspace(-2*np.pi, 2*np.pi, 100)

# y 데이터 생성 (sin(x))
y = np.sin(x)

# 첫 번째 그래프 그리기 (빨간색, 삼각형 마커, 점선)
# label은 범례에 표시될 이름
plt.plot(x, y, 'r:^', label='legend1')

# 두 번째 그래프 그리기 (파란색, 원 마커, 점선)
plt.plot(x, 0.5*y, color='blue', marker='o', linestyle=':',label='legend2')

# 그래프 제목
plt.title('Sine graph')

# x축 레이블 (LaTeX 스타일)
plt.xlabel(r'$x$')

# y축 레이블 (LaTeX 스타일)
plt.ylabel(r'$sin(x)$')

# 그리드 표시
plt.grid(True)

# 범례 표시 (왼쪽 위에 위치)
plt.legend(loc='upper left')

# 그래프 출력
plt.show()
```

```python
np.testing.assert_allclose(y, np.sin(x))
print("예제 계산 확인")
```

## Subplot · 감쇠하는 cosine

y1과 y2의 차이를 식에서 설명하고 x가 커질 때 진폭을 예측.

# Subplot


```python
# 필요한 라이브러리 임포트
import matplotlib.pyplot as plt
import numpy as np

# x 데이터 생성 (0부터 1까지 50개의 점)
x = np.linspace(0, 1, 50)

# 첫 번째 y 데이터 생성 (cos(4πx))
y1 = np.cos(4 * np.pi * x)

# 두 번째 y 데이터 생성 (cos(4πx) * exp(-2x))
y2 = np.cos(4 * np.pi * x) * np.exp(-2 * x)

# 첫 번째 그래프 그리기 (2행 1열, 첫 번째 위치)
plt.subplot(2, 1, 1)
plt.plot(x, y1, 'r-*', lw=1)  # 빨간색, 별 마커, 실선, 선 굵기 1
plt.grid(True)                # 그리드 표시
plt.ylabel(r'$cos(4 \pi x)$') # y축 레이블 설정 (LaTeX 스타일)

# 두 번째 그래프 그리기 (2행 1열, 두 번째 위치)
plt.subplot(2, 1, 2)
plt.plot(x, y2, 'bo', lw=1)   # 파란색, 원 마커, 실선, 선 굵기 1
plt.grid(True)                # 그리드 표시
plt.xlabel('x')               # x축 레이블 설정
plt.ylabel(r'$ e^{-2x} cos(4\pi x) $') # y축 레이블 설정 (LaTeX 스타일)

# 그래프 출력
plt.show()
```

```python
np.testing.assert_allclose(y2, y1 * np.exp(-2*x))
print("예제 계산 확인")
```

## 1. 상자그림


```python
# 필요한 라이브러리 임포트
import matplotlib.pyplot as plt # 그래프 그리기 라이브러리
import numpy as np # 수치 계산 라이브러리

# 데이터 생성 (평균 0, 표준편차 4인 정규분포에서 100개의 샘플)
data = np.random.normal(0, 4, 100)

# 상자 그림 그리기
a = plt.boxplot(data, orientation="horizontal") # orientation="horizontal": 가로 상자 그림
plt.title('Basic Plot') # 그래프 제목 설정

# 그래프 출력
plt.show()
```

```python
# 필요한 라이브러리 임포트
import seaborn as sns
import matplotlib.pyplot as plt

# 붓꽃 데이터 로드
iris = sns.load_dataset("iris", cache=False)

# 꽃받침 길이와 꽃받침 너비 데이터 추출
s1 = iris['sepal_length'].to_list() # 꽃받침 길이를 리스트로 변환
s2 = iris['sepal_width'].to_list()  # 꽃받침 너비를 리스트로 변환

# 상자 그림 그리기
plt.boxplot([s1, s2])     # s1과 s2 데이터를 사용하여 상자 그림 생성
plt.title('Iris data')    # 제목
plt.xticks([1, 2], ['sepal_length', 'sepal_width']) # x축 눈금 레이블 설정

# 그래프 출력
plt.show()
```

## 2. 히스토그램


```python
plt.hist([s1, s2], label=['sepal_length', 'sepal_width'])
plt.legend() # 범례 표시

# 그래프 출력
plt.show()
```

## 3. 산점도


```python
plt.scatter(s1, s2, color='r')
plt.ylabel("sepal_width")       # y축 레이블 설정
plt.xlabel("sepal_length")      # x축 레이블 설정

# 그래프 출력
plt.show()
```

```python
# 붓꽃 종류별로 데이터를 그룹화
groups = iris.groupby('species')

# 그래프 생성
fig, ax = plt.subplots()

# 각 그룹에 대해 산점도 그리기
for name, group in groups:
    ax.scatter(group.petal_length,  # x축: 꽃잎 길이
               group.petal_width,   # y축: 꽃잎 너비
               marker='o',          # 마커 모양: 원
               label=name)          # 범례 레이블: 종 이름

# 범례
ax.legend(fontsize=12, loc='upper left')

# 그래프 제목
plt.title('Scatter Plot by matplotlib', fontsize=15)

# x축 레이블
plt.xlabel('Petal Length', fontsize=12)

# y축 레이블
plt.ylabel('Petal Width', fontsize=12)

# 그래프 출력
plt.show()
```

## 3. 막대그래프


```python
# 데이터 설정
labels = ['G1', 'G2', 'G3', 'G4', 'G5'] # 막대 그래프의 레이블
means = [20, 34, 30, 35, 27]            # 각 레이블에 해당하는 값

# 막대 그래프 그리기
x = np.arange(len(labels))          # 레이블 위치 (0, 1, 2, 3, 4)
rec = plt.bar(x, means, 0.5)        # 막대 그래프 생성, 막대 두께 0.5
plt.xticks(x, labels)               # x축 눈금 레이블
plt.title("bar plot", fontsize=12)  # 그래프 제목, 폰트 크기 12

# 그래프 출력
plt.show()
```

## 4. 원그래프


```python
# 파이 차트 데이터 설정
labels = ['Frogs', 'Hogs', 'Dogs', 'Logs']  # 각 조각의 레이블
sizes = [15, 30, 45, 10]                    # 각 조각의 크기 (비율)
explode = (0, 0.1, 0, 0)                    # 두 번째 조각 ('Hogs')만 0.1만큼 돌출

# 파이 차트 그리기
x = plt.pie(sizes,              # 조각 크기 데이터
            explode=explode,    # 조각 돌출 설정
            labels=labels,      # 조각 레이블
            autopct='%d',       # 백분율 표시 형식 (정수)
            shadow=True,        # 그림자 효과 추가
            startangle=90)      # 첫 번째 조각의 시작 각도 (90도)

# 그래프 출력
plt.show()
```

## boxplot 비교

사각형·notch 상자그림과 species별 상자그림에서 비교 단위 구분.

# 라이브러리별 boxplot 비교


## 1. pandas


```python
import matplotlib.pyplot as plt
import numpy as np

# 랜덤 테스트 데이터 생성
np.random.seed(19680801) # 재현성을 위한 랜덤 시드 설정
all_data = [np.random.normal(0, std, size=100) for std in range(1, 4)] # 표준 편차가 다른 세 개의 데이터셋 생성
labels = ['x1', 'x2', 'x3'] # 데이터셋 레이블

# 그래프와 축 생성
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(9, 4)) # 1행 2열, 크기 (9, 4)인 그래프 생성

# 사각형 상자 그림
bplot1 = axes[0].boxplot(all_data,          # 데이터
                         orientation="vertical",         # 세로 방향 정렬
                         patch_artist=True, # 색상 채우기
                         tick_labels=labels)     # x축 눈금 레이블
axes[0].set_title('사각형 상자 그림')       # 첫 번째 subplot 제목 설정

# 노치 모양 상자 그림
bplot2 = axes[1].boxplot(all_data,          # 데이터
                         notch=True,        # 노치 모양
                         orientation="vertical",         # 세로 방향 정렬
                         patch_artist=True, # 색상 채우기
                         tick_labels=labels)     # x축 눈금 레이블
axes[1].set_title('노치 모양 상자 그림')    # 두 번째 subplot 제목 설정

# 상자 색상 채우기
colors = ['pink', 'lightblue', 'lightgreen']          # 상자 색상 정의
for bplot in (bplot1, bplot2):                        # 두 상자 그림에 대해 반복
    for patch, color in zip(bplot['boxes'], colors):  # 상자와 색상 반복
        patch.set_facecolor(color)                    # 상자 색상 설정

# 가로 그리드 선 및 레이블 추가
for ax in axes:                     # 두 축에 대해 반복
    ax.yaxis.grid(True)             # 가로 그리드 선 추가
    ax.set_xlabel('세 개의 샘플')   # x축 레이블 설정
    ax.set_ylabel('관측값')         # y축 레이블 설정

# 그래프 출력
plt.show()
```

```python
# 종별로 그룹화된 상자 그림 그리기
iris.boxplot(by='species') # 'species' 열을 기준으로 데이터를 그룹화하여 상자 그림 생성

# 그래프 출력
plt.show()
```
