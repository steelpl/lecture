## 식 계산

2+3, 2*3, 2**3의 결과를 먼저 예상한 뒤 실행하시오.

```python
print(2 + 3)
print(2 * 3)
print(2 ** 3)
```

## 코드 한 부분 바꾸기

아래 코드는 4가 목록 안에 있을 때 글자를 출력한다. 첫 줄의 4를 5로 바꾸어 결과를 비교하시오.

```python
if 4 in [1, 2, 3, 4]:
    print("4가 있습니다")
```

들여쓴 줄은 조건이 맞을 때 실행. 조건이 틀리면 이 셀에서는 출력이 없음. 조건문의 자세한 구조는 제어문에서 학습.
