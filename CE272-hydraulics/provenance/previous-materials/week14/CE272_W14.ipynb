## Froude 모형과 동시 상사의 한계

**먼저 예측**: 같은 물을 쓰는 축소 모형에서 Froude 수와 Reynolds 수를 동시에 같게 할 수 있는가?

**문제**: 길이 500 m인 원형 수로를 10:1 축척으로 50 m 모형으로 제작한다. 원형 유량이 400 m³/s일 때 Froude 모형법칙의 모형 유량과 힘비를 구하시오. 같은 물을 쓸 때 Reynolds 수 비도 계산하시오.

예상값·단위·식·가정을 먼저 기록한 뒤 아래 계산과 비교


```python
Lr,Qp=10.,400.
ratios=hyd.model_ratios(Lr,np.sqrt(Lr))
pd.Series({**ratios,'Qm_m3_s':Qp/ratios['Qr'],
           'Re_prototype_model_same_water':ratios['Vr']*Lr,
           'required_nu_prototype_model_for_both':Lr**1.5})
```

## 왜곡모형과 Manning 조도비

**먼저 예측**: 수평과 수직 축척이 다르면 기하학적으로 닮은 모형인가?

**문제**: 수평 축척 Xr=300, 수직 축척 Yr=60을 추가로 가정한다. 원본의 광폭 직사각형·Froude 가정 아래 속도비, 시간비, 유량비와 Manning n비를 계산하시오. 왜곡비와 가정의 한계를 설명하시오.

예상값·단위·식·가정을 먼저 기록한 뒤 아래 계산과 비교


```python
Xr,Yr=300.,60.
Vr=Yr**.5;Tr=Xr/Vr;Qr=Xr*Yr**1.5
nr=Yr**(2/3)/Xr**.5
pd.Series({'Xr':Xr,'Yr':Yr,'distortion_Xr_Yr':Xr/Yr,
           'Vr':Vr,'Tr':Tr,'Qr':Qr,'n_prototype_model':nr,'bed_slope_ratio':Yr/Xr})
```
