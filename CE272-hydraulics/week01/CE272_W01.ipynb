**먼저 예측**: 물 1 m³의 질량과 중량은 각각 어떤 단위와 크기인가?

**문제**: 밀도 1,000 kg/m³인 물 1 m³의 질량과 중량을 계산하시오. 수심 3 m의 게이지압력을 계산하고 kg, kgf, N, Pa의 차이를 설명하시오.

예상값·단위·식·가정을 먼저 기록한 뒤 아래 계산과 비교

```python
rho, volume, g, depth = 1000., 1., hyd.G, 3.
mass = rho * volume
weight = mass * g
p_gauge = rho * g * depth
pd.Series({'mass_kg':mass, 'weight_N':weight, 'gauge_pressure_Pa':p_gauge,
           'gauge_pressure_kPa':p_gauge/1000})
```

예상값·단위·식·가정을 먼저 기록한 뒤 아래 계산과 비교

```python
b, h, rho = 2., 3., 1000.
depths = np.linspace(0,h,1001)
pressures = rho*hyd.G*depths
force_formula = rho*hyd.G*b*h**2/2
force_integral = b*trapezoid(pressures,depths)
pd.Series({'force_N':force_formula, 'integral_N':force_integral,
           'center_of_pressure_depth_m':2*h/3, 'difference_N':force_integral-force_formula})
```
