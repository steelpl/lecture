## 값·자료형·계산

강우량 `[0, 20, 5] mm`의 합계와 최댓값을 구하시오. 문자열 `"20"`과 숫자 `20`의 차이를 설명하시오.

```python
rain = [0, 20, 5]
print(sum(rain), max(rain))
text = '20'
print(text + text, int(text) + int(text))
print(type(rain), type(rain[0]), rain[1:])
assert sum(rain) == 25
```

## 배열과 결측

`np.mean`과 `np.nanmean`의 결과를 비교하고, 결측 표시를 `-999`로 바꾸기만 하면 어떤 문제가 생기는지 설명하시오.

```python
a = np.array([1.2, np.nan, 1.8])
print(np.mean(a), np.nanmean(a))
bad = np.array([1.2, -999, 1.8])
print('결측 부호를 숫자로 계산:', bad.mean())
```
