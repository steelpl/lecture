## 값과 계산

강우량 [0,20,5] mm의 합계와 최댓값을 구하시오. 숫자 20과 문자열 "20"의 덧셈 결과를 구분하시오.

```python
rain=[0,20,5]
print(sum(rain),max(rain))
print(20+20)
print("20"+"20")
```
