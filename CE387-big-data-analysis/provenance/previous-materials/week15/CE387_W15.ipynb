```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
np.set_printoptions(precision=5, suppress=True)
```

## 분석 계획

수위 `[1.2,1.4,결측,1.8] m`가 주어졌다. 결측 처리, 요약값, 단위, 해석 범위를 먼저 정하시오. 코드 실행 후 계획과 일치하는지 검토하시오.

```python
water=pd.Series([1.2,1.4,np.nan,1.8])
print({'observed':int(water.count()),'missing':int(water.isna().sum()),
       'mean_m':water.mean(),'maximum_m':water.max()})
assert water.count()==3
```

## 모형의 결과 검토

붓꽃 분류 자료를 훈련·시험으로 나누시오. 전처리를 훈련 자료에만 맞추고, 정확도와 혼동행렬을 함께 제시하시오.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix
X,y=load_iris(return_X_y=True)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,stratify=y,random_state=42)
pipe=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000))
pipe.fit(Xtr,ytr);pred=pipe.predict(Xte)
print('accuracy:',accuracy_score(yte,pred));print(confusion_matrix(yte,pred))
assert len(Xtr)+len(Xte)==150
```
