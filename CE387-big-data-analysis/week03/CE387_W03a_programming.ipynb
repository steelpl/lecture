```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
np.set_printoptions(precision=5, suppress=True)
```

## 제어문

강우량이 10 mm 이상인 날의 수와 해당 값만의 합계를 구하시오. 10 mm를 포함하는 조건인지 확인하시오.

```python
rain = [0, 20, 5, 10]
heavy = []
for value in rain:
    if value >= 10:
        heavy.append(value)
print(heavy, len(heavy), sum(heavy))
assert heavy == [20, 10]
```

## 함수

평균을 계산하는 함수를 작성하시오. 빈 자료가 들어왔을 때의 처리도 정하시오.

```python
def mean_rain(values):
    if len(values) == 0:
        raise ValueError('자료가 비어 있음')
    return sum(values) / len(values)
print(mean_rain(rain))
try:
    mean_rain([])
except ValueError as error:
    print(error)
```

## 라이브러리

반복문으로 구한 결과를 NumPy·pandas 결과와 비교하시오. 임계값을 5 mm로 바꾸고 세 방법이 계속 일치하는지 확인하시오.

```python
a = np.array(rain)
s = pd.Series(rain)
print(a[a >= 10], s[s >= 10].tolist())
assert np.array_equal(a[a >= 10], heavy)
assert np.isclose(a.mean(), mean_rain(rain))
plt.bar(range(1, len(rain)+1), rain)
plt.axhline(10, color='red', ls='--')
plt.xlabel('Day'); plt.ylabel('Rainfall (mm)'); plt.show()
```
