```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
np.set_printoptions(precision=5, suppress=True)
```

## 직선의 기울기·절편

x=[10,20,…,80], y=[25,70,380,550,610,1220,830,1450]의 최소제곱 직선을 구하시오. 중간 반올림 없이 계산하고 잔차합을 확인하시오.

```python
x=np.arange(10.,81.,10.); y=np.array([25.,70.,380.,550.,610.,1220.,830.,1450.])
slope=((x-x.mean())@(y-y.mean()))/np.sum((x-x.mean())**2)
intercept=y.mean()-slope*x.mean()
error=y-(slope*x+intercept)
print('기울기·절편:',slope,intercept,'잔차합:',error.sum())
assert np.isclose(error.sum(),0)
plt.scatter(x,y);plt.plot(x,slope*x+intercept);plt.xlabel('x');plt.ylabel('y');plt.show()
```

## 붓꽃 분류

꽃잎 너비·꽃받침 너비 두 변수로 세 종을 분류하시오. 분할→훈련 자료 기준 표준화→학습→시험 자료 평가 순서로 진행하시오.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score
iris=load_iris(); X=iris.data[:,[3,1]];target=iris.target
Xtr,Xte,ytr,yte=train_test_split(X,target,test_size=.25,stratify=target,random_state=42)
model=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000))
model.fit(Xtr,ytr);pred=model.predict(Xte)
print('시험 정확도:',accuracy_score(yte,pred))
display(pd.DataFrame(confusion_matrix(yte,pred),index=iris.target_names,columns=iris.target_names))
print('첫 시험 표본의 확률:',model.predict_proba(Xte[:1]))
```
