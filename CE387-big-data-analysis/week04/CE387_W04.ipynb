## 조건부확률

검사 전 감염확률 0.1%, 민감도 90%, 특이도 99%인 가상의 검사 상황이다. 양성 판정을 받은 사람의 감염확률을 구하시오.

```python
prior = 0.001
sensitivity = 0.90
specificity = 0.99
true_positive = prior * sensitivity
false_positive = (1-prior) * (1-specificity)
posterior = true_positive / (true_positive + false_positive)
print('양성 중 감염 확률:', posterior)
people = 100_000
print('가상 인원:', {'감염·양성':people*true_positive,
                    '비감염·양성':people*false_positive})
assert np.isclose(posterior, 90/1089)
```

풀이: 감염·양성 90명, 비감염·양성 999명 → 90/1,089 ≈ 8.26%. 민감도 90%와 서로 다른 조건부확률.

사전확률만 1%, 10%로 바꾸어 비교하시오. 검사 성능이 같아도 결과가 달라지는 이유를 설명하시오.

```python
rows = []
for p in [0.001, 0.01, 0.1]:
    posterior = p*sensitivity / (p*sensitivity + (1-p)*(1-specificity))
    rows.append([p, posterior])
display(pd.DataFrame(rows, columns=['prior','P(infected | positive)']))
```

## 순차 갱신

가상의 두 상태 A·B의 사전확률은 각각 0.5이다. 세 관찰의 가능도쌍은 `(0.4,0.3)`, `(0.2,0.01)`, `(0.7,0.9)`이다. 조건부독립을 가정하여 관찰마다 A의 확률을 갱신하시오.

```python
p = 0.5
for likelihood_a, likelihood_b in [(0.4,0.3),(0.2,0.01),(0.7,0.9)]:
    p = p*likelihood_a / (p*likelihood_a+(1-p)*likelihood_b)
    print(f'{p:.6f}')
assert np.isclose(p, 0.028/(0.028+0.00135))
```
