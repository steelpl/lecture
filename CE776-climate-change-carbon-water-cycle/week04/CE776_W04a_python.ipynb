## 식과 변수

`=`는 오른쪽 계산 결과를 왼쪽 이름에 저장하는 연산. 셀은 같은 실행 상태를 공유

```python
a = 3
b = 4
print(a + b, a * b, a / b)
print(type(a), type(a / b))
```

`a`를 6으로 바꾸고 이 셀을 다시 실행하시오. 나눗셈 결과와 자료형을 설명하시오.

```python
text = "Life is too short, You need Python"
print(text[0], text[-1])
print(text[0:4])
print(text.split(","))
```

## 여러 값에서 필요한 자료 선택

인덱스는 0부터, 슬라이스의 끝 위치는 제외

```python
values = [1, 2, 3, ["a", "b", "c"]]
print(values[0], values[-1])
print(values[1:3])
print(values[3][1])
```

문자열의 `text[0:4]`와 리스트의 `values[1:3]`을 비교하시오. 끝 위치를 포함한다고 생각하면 어떤 오류가 생기는지 설명하시오.
