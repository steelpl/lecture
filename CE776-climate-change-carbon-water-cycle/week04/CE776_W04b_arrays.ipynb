# 반복·함수·배열

여러 자료에 같은 계산을 적용하고 모양을 확인

**할 일** · 짝수의 제곱·5×5 배열·함수 반환값 비교

위에서 아래로 실행. 코드를 바꾸기 전 결과를 예상하고, 실행 결과와 비교

```python
squares = []
for x in range(1, 11):
    if x % 2 == 0:
        squares.append(x ** 2)
print(squares)
print(list(range(2, 12, 2)))
```

결과 `[4, 16, 36, 64, 100]`을 확인하시오. `range(1, 11)`을 `range(1, 10)`으로 바꾸면 무엇이 달라지는지 설명하시오.

```python
import numpy as np
identity = np.eye(5)
array = np.arange(1, 26).reshape(5, 5)
print(identity)
print(array)
print("shape:", array.shape)
print("first row:", array[0, :])
print("column means:", array.mean(axis=0))
```

`axis=0`과 `axis=1`의 평균을 비교하시오. 무엇을 따라 평균을 계산했는지 원래 배열에서 확인하시오.

```python
def add(a, b):
    return a + b

result = add(3, 4)
print(result)
print(add(b=4, a=3))
```

**정리** · 반복은 계산의 재사용, 함수는 입력–출력의 분리, 배열은 모양과 축을 갖는 자료. 위성영상도 밴드·행·열로 이루어진 배열
