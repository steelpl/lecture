# Loops, functions and arrays

Apply the same calculation to several values and inspect the shape.

**Task** · Compare even-number squares, a 5×5 array and function return values.

Run from top to bottom. Predict the result before changing code, then compare it with the output.

```python
squares = []
for x in range(1, 11):
    if x % 2 == 0:
        squares.append(x ** 2)
print(squares)
print(list(range(2, 12, 2)))
```

Check the result `[4, 16, 36, 64, 100]`. Explain what changes when `range(1, 11)` becomes `range(1, 10)`.

```python
import numpy as np
identity = np.eye(5)
array = np.arange(1, 26).reshape(5, 5)
print(identity)
print(array)
print("shape:", array.shape)
print("first row:", array[0, :])
print("column means:", array.mean(axis=0))
```

Compare means with `axis=0` and `axis=1`. Use the original array to explain which values are averaged.

```python
def add(a, b):
    return a + b

result = add(3, 4)
print(result)
print(add(b=4, a=3))
```

## Transpose and dot product

Transpose the 3×2 matrix and calculate the dot product of the two vectors. Predict the shapes and results before running the code.

```python
matrix = np.array([[1, 2], [3, 4], [5, 6]])
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("transpose:\n", matrix.T)
print("dot product:", np.dot(a, b))
```

The transpose has shape 2×3; the dot product is 32. Transposing preserves the number and values of elements. A dot product multiplies corresponding elements and sums them to give a scalar.

**Takeaway** · Loops reuse calculations, functions separate inputs from outputs, and arrays have shapes and axes. Satellite images are also arrays with bands, rows and columns.
